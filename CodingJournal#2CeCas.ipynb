```python
import matplotlib.pyplot as plt 
import numpy as np 
```

```python
x=np.linspace(0,2*np.pi,100)
a=55.29683
b=27.92451
def exponential_function(x,a,b):
    return a*np.exp(-b*x)
```

```python
y=a*np.exp(-b*x)
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Exponential Function')
plt.savefig("ExponentialFunction.pdf")
plt.show()
```

```python
sigma=np.linspace(2,0.2,len(x))
np.random.normal(0,sigma)
y_scatter= y+np.random.normal(0,sigma)

plt.errorbar(x,y_scatter, yerr=sigma,fmt="*",color="Pink")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.savefig("ScatterPlot.pdf")
plt.show()

```

# Prompt #3

```python
from scipy.optimize import curve_fit

def model_exponential(x,a,b):
    return  a*np.exp(-b*x)
def model_function(x,a,b):
    return  a/((x+10**-15)**b) ##added 10^-15 so the denominator does not reach zero
popt_exp, _ = curve_fit(model_exponential, x, y_scatter)
popt_power, _ = curve_fit(model_function, x, y_scatter)

print(f"Exponential Model Fit: a={popt_exp[0]}, b={popt_exp[1]}")
print(f"Function Model Fit: a={popt_power[0]}, b={popt_power[1]}")

plt.plot(x, model_exponential(x, *popt_exp), color="purple", label="Exponential Fit")
plt.plot(x, model_function(x, *popt_power), color="magenta", label="Function Fit")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Best-Fit Plot')
plt.savefig("BestFitPlot.pdf")
plt.legend()
plt.show()



```

```python
import numpy as np
import matplotlib.pyplot as plt

scale_parameter = 5.0
random_numbers = np.random.exponential(scale=scale_parameter, size=1000)

plt.hist(random_numbers, bins=100, color='cyan', edgecolor='black', alpha=0.5)


plt.xlabel('Value')
plt.ylabel('Frequency (Count)')
plt.title('Histogram of Exponential Distribution')
plt.savefig("HistogramOfExponentialDistribution.pdf")
plt.show()
```
